```python
# Version simple pour juste le graphique du nombre de portes(outliers manitha fams des voitures les prix mte3hom anormale )
plt.figure(figsize=(12, 8))

if 'Nombre_portes' in df.columns:
    # Boxplot amélioré
    sns.boxplot(x='Nombre_portes', y='Price', data=df,
                palette='Set2',
                showmeans=True,
                meanprops={"marker":"o",
                          "markerfacecolor":"white",
                          "markeredgecolor":"black",
                          "markersize":"8"})

    plt.title('Distribution du Prix par Nombre de Portes',
              fontsize=16, fontweight='bold', pad=20)
    plt.xlabel('Nombre de Portes', fontsize=14, fontweight='bold', labelpad=10)
    plt.ylabel('Prix (DT)', fontsize=14, fontweight='bold', labelpad=10)

    # Ajouter des annotations pour les moyennes
    for i, n_portes in enumerate(sorted(df['Nombre_portes'].dropna().unique())):
        subset = df[df['Nombre_portes'] == n_portes]
        mean_price = subset['Price'].mean()
        count = subset.shape[0]

        plt.text(i, mean_price + (df['Price'].max() * 0.02),
                 f'Moyenne: {mean_price:,.0f} DT\n(n={count})',
                 ha='center', va='bottom',
                 fontsize=10, fontweight='bold',
                 bbox=dict(boxstyle="round,pad=0.3", facecolor="white", edgecolor="gray", alpha=0.8))

    # Grille
    plt.grid(True, axis='y', alpha=0.3, linestyle='--')
    plt.gca().set_facecolor('#f8f9fa')

plt.tight_layout()
plt.show()
```

```python
# Prix moyen par puissance
if 'Puissance_ch' in df.columns:

    plt.figure(figsize=(16, 8))

    top_5_puissances = df['Puissance_ch'].value_counts().head(5).index
    subset = df[df['Puissance_ch'].isin(top_5_puissances)]

    pivot_data = subset.pivot_table(
        values='Price',
        index='Puissance_ch',
        aggfunc='mean'
    )

    pivot_data.plot(kind='bar', color='skyblue')

    plt.title("Prix Moyen par Puissance", fontsize=14, fontweight='bold')
    plt.xlabel('Puissance (ch)')
    plt.ylabel('Prix Moyen (DT)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Prix moyen par nombre de cylindres
if 'Cylindree' in df.columns:

    plt.figure(figsize=(16, 8))

    top_5_cylindres = df['Cylindree'].value_counts().head(5).index
    subset = df[df['Cylindree'].isin(top_5_cylindres)]

    pivot_data = subset.pivot_table(
        values='Price',
        index='Cylindree',
        aggfunc='mean'
    )

    pivot_data.plot(kind='bar', color='lightgreen')

    plt.title("Prix Moyen par Nombre de Cylindres", fontsize=14, fontweight='bold')
    plt.xlabel('Nombre de Cylindres')
    plt.ylabel('Prix Moyen (DT)')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()



```

```python
# nettoyer les espaces dans les colonnes numériques
cols = ['Kilometrage','Puissance_fiscale','Puissance_ch','Cylindree']

for col in cols:
    df[col] = df[col].astype(str).str.replace(' ', '')
```

```python
#convertir en numérique (en sécurité)
for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
#verfication
df[cols].info()
```

```python
import seaborn as sns
# Calcul de la corrélation
corr = df[cols].corr()

# Création de la heatmap
plt.figure(figsize=(10, 8))  # taille de la figure
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Matrice de corrélation")
plt.show()
```
